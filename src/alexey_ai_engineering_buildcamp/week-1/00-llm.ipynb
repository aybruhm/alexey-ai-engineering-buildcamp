```python
from anthropic import Anthropic
```

```python
client = Anthropic()
```

```python
messages = client.messages.create(
    model="claude-opus-5",
    messages=[
        {
            "role": "user",
            "content": "Write a poem about the beauty of nature.",
        }
    ],
    max_tokens=1024,
)

print("Response JSON: ", messages.model_dump_json(indent=2))
print("\n\n")
for block in messages.content:
    if block.type =="text":
        print(block.text)
```

```python
with client.messages.stream(
    model="claude-opus-5",
    messages=[
        {
            "role": "user",
            "content": "Write a poem about the beauty of universe.",
        }
    ],
    max_tokens=2056,
) as stream:
    for text in stream.text_stream:
        print(text, flush=True, end="")
```

```python
MODEL_PRICING = {
    "claude-opus-5": {"input": 5.00, "output": 25.00},
    "claude-sonnet-5": {"input": 3.00, "output": 15.00},
    "claude-haiku-4.5": {"input": 1.00, "output": 5.00},
}

def calculate_cost(model_name: str, input_tokens: int, output_tokens: int) -> float:
    if model_name not in MODEL_PRICING:
        raise ValueError(f"Model {model_name} not found in pricing.")
    
    pricing = MODEL_PRICING[model_name]
    input_cost = (input_tokens / 1_000_000) * pricing["input"]
    output_cost = (output_tokens / 1_000_000) * pricing["output"]
    
    return input_cost + output_cost
```

```python
usage = messages.usage
total_cost = calculate_cost(messages.model, usage.input_tokens, usage.output_tokens)
print(f"Total cost for the request: ${total_cost:.4f}")
```
